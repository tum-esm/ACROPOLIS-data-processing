```python
1e-3 / 60 / 2 # 0.5 L/min to m^3/s
```

```python
%%latex
$$
1~\mathrm{L\,min^{-1}} = \frac{1\times10^{-3}}{60}\mathrm{m^3\,s^{-1}}\ \mathrm{m^3\,s^{-1}} = 1.667\times10^{-5}\ \mathrm{m^3\,s^{-1}}
$$

$$
Q = 0.5~\mathrm{L\,min^{-1}} = 0.5\times1.667\times10^{-5}\ \mathrm{m^3\,s^{-1}}= 8.33\times10^{-6}~\mathrm{m^3\,s^{-1}}
$$
```

```python
%%latex
Given:

\begin{aligned}
&\text{Tube inner diameter:} & D &= 4.3\ \mathrm{mm} = 4.3\times 10^{-3}\ \mathrm{m} \\
&\text{Length:} & L &= 50\ \mathrm{m} \\
&\text{Volumetric flow:} & Q &= 0.5\ \mathrm{L\,min^{-1}} = 8.333\times 10^{-6}\ \mathrm{m^3\,s^{-1}} \\
&\text{Air at } \approx 20^\circ\mathrm{C:} & \mu &\approx 1.85\times 10^{-5}\ \mathrm{Pa\,s},\quad \rho \approx 1.2\ \mathrm{kg\,m^{-3}}
\end{aligned}
```

```python
A = np.pi * (0.0043)**2 / 4
A
```

```python
%%latex
Cross-sectional area
$$
A=\frac{\pi D^2}{4}=\frac{\pi (0.0043)^2}{4}\ \mathrm{m^2}=1.452\times10^{-5}\ \mathrm{m^2}.
$$
```

```python
v = 8.333e-6 / A
v
```

```python
%%latex
Mean velocity
$$
v=\frac{Q}{A}=\frac{8.333\times10^{-6}}{1.452\times10^{-5}}\ \mathrm{m\,s^{-1}}=0.574\ \mathrm{m\,s^{-1}}.
$$
```

```python
t = 10 / v
t
```

```python
t = 20 / v
t
```

```python
t = 50 / v
t / 60
```

```python
%%latex
Residence time (a.k.a. advective delay)
$$
t=\frac{L}{v}=\frac{50}{0.574}\ \mathrm{s}=87.1\ \mathrm{s}=1.45\ \mathrm{min}.
$$
```

```python
1.2 * v * 0.0043 / 1.85e-5
```

```python
%%latex
Reynolds number (laminar vs. turbulent check)
$$
\mathrm{Re}=\frac{\rho v L}{\mu}=\frac{(1.2)(0.574)(0.0043)}{1.85\times10^{-5}} \approx 160\ (<2000\ \Rightarrow\ \text{laminar}).
$$
with $\rho$ being the fluid density, $v$ the flow velocity, $L$ the characteristic length (here: tube diameter), and $\mu$ the dynamic viscosity.
```

```python
delta_p = (128 * 1.85e-5 * 50 * 8.333e-6) / (np.pi * (0.0043)**4)
delta_p / 100
```

```python
%%latex
Laminar pressure drop (Hagen-Poiseuille)
$$
\Delta p=\frac{8\,\mu\,L\,Q}{\pi R^4}
=\frac{128\,\mu\,L\,Q}{\pi D^4}
=\frac{128(1.85\times10^{-5})(50)(8.333\times10^{-6})}{\pi(0.0043)^4}\ \mathrm{Pa}
\approx 9.19\times10^{2}\ \mathrm{Pa}=0.92\ \mathrm{kPa}.
```
