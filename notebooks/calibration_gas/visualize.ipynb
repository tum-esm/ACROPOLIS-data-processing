```python
import polars as pl
import numpy as np
import matplotlib.pyplot as plt
import os

PROJECT_PATH = os.path.abspath(os.path.join("..", ".."))
DATA_DIRECTORY = os.path.join(PROJECT_PATH, "data")
```

```python
df = pl.read_csv(os.path.join(DATA_DIRECTORY, "input", "averaged_gases.csv")).filter(pl.col("cal_bottle_id") > 40)
df_400 = df.filter(pl.col("cal_bottle_CO2") < 450)
df_520 = df.filter(pl.col("cal_bottle_CO2") > 450)
```

```python
y = [1 for _ in range(len(df_520))]
plt.scatter(df_520["cal_bottle_CO2"], y=y, s=0.7)
plt.show()
plt.hist(df_520["cal_bottle_CO2"], bins=100, alpha=0.5)
plt.show()
```
